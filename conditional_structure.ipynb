```python
# Enter a number to check sum of positive integer
# 1 to 10 -even -2,4,6,8,10 Sum()=30
num=int(input("enter the number"))
total_sum = 0
i = 1
if num%2==0:
    while i <= num:
        total_sum += i
        i += 1
        print(f"The sum of {num} even number is{total_sum}")
else:
    while i <= num:
        total_sum += i
        i += 1
        print(f"The sum of {num} odd number is{total_sum}")
```
