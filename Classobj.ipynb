```python

class AirlineTicket:

    def display_ticket_details(self, name, age, source, destination):
        print(f"Passenger: {name} of age: {age} is travelling from: {source} to: {destination}")

    def confirm_ticket(self):
        return "Confirmed"

    def cancel_ticket(self):
        return "Cancelled"

    def update_destination(self, old_destination, new_destination):
        print(f"Destination updated from {old_destination} to {new_destination}")
        return new_destination


# Create an object
ticket = AirlineTicket()

# 1. Display Ticket Details
ticket.display_ticket_details("Sathya", 25, "Chennai", "Delhi")

# 2. Confirm Ticket
print("Ticket Status:", ticket.confirm_ticket())

# 3. Cancel Ticket
print("Ticket Status:", ticket.cancel_ticket())

# 4. Update Destination
updated_destination = ticket.update_destination("Delhi", "Mumbai")
print("New Destination:", updated_destination)
```
