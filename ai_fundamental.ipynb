```python
class Telephone:
    def __init__(self, model, costperhour):
        self.model = model
        self.costperhour = costperhour

    def installationcost(self, hours):
        return hours * self.costperhour

# Instantiating Object
r = Telephone("rotary", 50)
print(r.installationcost(10))
```
