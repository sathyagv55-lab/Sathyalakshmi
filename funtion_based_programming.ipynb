```python
#2. Calling Functions - Positional & Keyword Arguments
#AI says, Parameter (named) = Placeholder, Argument(positition) = actual value
def calculate_offer(amt,off_pct,shipping_charges):
    offer_applied_amt=amt-(amt*off_pct)
    total_bill_amt=offer_applied_amt+shipping_charges
    return total_bill_amt

#Positional Argument Function Call
result=calculate_offer(1000,.05,100)#positional arguments (arguments)#Internally amt=1000, off_pct=.05, shipping_charges=100
print('Positional with right mapping',result)
result=calculate_offer(1000,100,.05)#Internally amt=1000, off_pct=100, shipping_charges=.05
print('Positional with wrong mapping',result)#wrong result if i use postitional argument

#Keyword/Named Argument Function Call
result=calculate_offer(amt=1000,off_pct=.05,shipping_charges=100)
print('Named is always right in mapping',result)
result=calculate_offer(amt=1000,shipping_charges=100,off_pct=.05)
print('Named is always right in mapping',result)#Right result if i use named/keyword parameter

#If we pass less or more params than expected
#more number of params/arguments fails
#result_more_arguments=calculate_offer(1000,.05,100,0)#errored out
#result_more_params=calculate_offer(amt=1000,off_pct=.05,shipping_charges=100,unknown=0)

#less number of params/arguments
#result_less_arguments=calculate_offer(1000,.05)#errored out
#result_less_arguments=calculate_offer(amt=1000,off_pct=.05)#errored out

#If I have chance of passing less number of arguments?
#We can go for DEFAULT ARGUMENT FUNCTIONS
def calculate_offer(amt,off_pct=.0,shipping_charges=0):
    offer_applied_amt=amt-(amt*off_pct)
    total_bill_amt=offer_applied_amt+shipping_charges
    return round(total_bill_amt)

#less number of params/arguments
result_less_arguments=calculate_offer(1000)#Default arguments works with positional..
print(result_less_arguments)

#We are using positional, named & default
def calculate_offer(amt,shipping_charges,off_pct=.0):
    offer_applied_amt=amt-(amt*off_pct)
    total_bill_amt=offer_applied_amt+shipping_charges
    return round(total_bill_amt)

result_less_arguments=calculate_offer(1000,shipping_charges=100)#Default arguments works..
print(result_less_arguments)

#result_less_arguments=calculate_offer(amt=1000,off_pct=.05)#Default arguments works..
```

```python
#4. Arbitrary (dynamic number of positional arguments) Argument Functions
def final_bill_fun(*args,shipping_charges):
    #Input parameter is considered as tuple type if we prefix * with the param name
    print(type(args))
    print("products amount",args)
    print("shipping Change",shipping_charges)
    total_amt=sum(args)
    final_bill_amt=total_amt-shipping_charges
    return final_bill_amt

total_amt=final_bill_fun(100,200,150,500,shipping_charges=50)
print(total_amt)
total_amt=final_bill_fun(10,40,20,shipping_charges=50)
print(total_amt)
```
